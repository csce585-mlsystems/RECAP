```python
import pandas as pd
import matplotlib.pyplot as plt

# Load training log
log_path = "../info/training_log.csv"
df = pd.read_csv(log_path)

df.head()
```

```python
fig, ax1 = plt.subplots(figsize=(8,5))

color = 'tab:red'
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Train Loss', color=color)
ax1.plot(df['epoch'], df['train_loss'], color=color, marker='o', label="Train Loss")
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()  # second axis for accuracy
color = 'tab:blue'
ax2.set_ylabel('Accuracy (%)', color=color)
ax2.plot(df['epoch'], df['train_acc'], color=color, marker='s', linestyle="--", label="Train Acc")
ax2.plot(df['epoch'], df['val_acc'], color='tab:green', marker='^', linestyle="--", label="Val Acc")
ax2.tick_params(axis='y', labelcolor=color)

fig.tight_layout()
fig.suptitle("Training Loss and Accuracy", y=1.05)
ax2.legend(loc="lower right")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['epoch'], df['val_macro_f1'], marker='o', color='purple')
plt.title("Validation Macro F1 over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Macro F1")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['epoch'], df['train_acc'], label="Train Acc", linestyle="--")
plt.plot(df['epoch'], df['val_acc'], label="Val Acc")
plt.plot(df['epoch'], df['val_macro_f1'], label="Val Macro F1")
plt.title("Accuracy & Macro-F1 over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Score")
plt.legend()
plt.grid(True)
plt.show()
```
